```python
import pandas as pd
import numpy as np
import joblib
import os
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingRegressor
from sklearn.svm import SVC
from sklearn.cluster import KMeans
from sklearn.neural_network import MLPClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Create models directory if it doesn't exist
if not os.path.exists('models'):
    os.makedirs('models')
    
print("Environment ready and 'models' directory created.")
```

## 1. Real Estate Price Predictor (Regression)
**Domain:** Real Estate / Finance  
**Description:** Predicts the market price of a house based on physical attributes.

### Schema
- **Input:** `[square_feet (int), num_bedrooms (int), house_age (int)]`
- **Output:** `[predicted_price (float)]`  
- **Input Shape:** `(1, 3)`
- **Output Shape:** `(1,)`

```python
# Synthetic Data
X = np.array([[1200, 2, 5], [2500, 4, 10], [800, 1, 20], [3000, 5, 2], [1500, 3, 15]])
y = np.array([250000, 500000, 150000, 700000, 300000])

# Model
model_1 = LinearRegression()
model_1.fit(X, y)

# Save
joblib.dump(model_1, 'models/house_price_model.pkl')

# Example usage
sample_input = [[1800, 3, 8]]
prediction = model_1.predict(sample_input)
print(f"Predicted Price for {sample_input}: ${prediction[0]:,.2f}")
```

## 2. Customer Churn Classifier (Binary Classification)
**Domain:** Telecom / CRM  
**Description:** Predicts if a customer will leave the service (Churn) based on usage and billing.

### Schema
- **Input:** `[tenure_months (int), monthly_charges (float), total_charges (float), has_internet (0 or 1)]`
- **Output:** `[churn_probability (float)]` or `[is_churn (int: 0/1)]`  
- **Input Shape:** `(1, 4)`
- **Output Shape:** `(1,)`

```python
# Synthetic Data
X = np.array([[1, 29.85, 29.85, 0], [34, 56.95, 1889.5, 1], [2, 53.85, 108.15, 1], [45, 42.3, 1840.75, 0], [10, 70.1, 700.5, 1]])
y = np.array([1, 0, 1, 0, 1])

# Model
model_2 = LogisticRegression()
model_2.fit(X, y)

# Save
joblib.dump(model_2, 'models/customer_churn_model.pkl')

# Example usage
sample_input = [[12, 65.0, 780.0, 1]]
prediction = model_2.predict(sample_input)
probability = model_2.predict_proba(sample_input)[0][1]
print(f"Churn Prediction: {'Yes' if prediction[0] == 1 else 'No'} (Prob: {probability:.2f})")
```

## 3. Iris Species Classifier (Multi-class Classification)
**Domain:** Biology / Botany  
**Description:** Identifies the species of an Iris plant based on flower measurements.

### Schema
- **Input:** `[sepal_length, sepal_width, petal_length, petal_width]`
- **Output:** `[species_id (0: Setosa, 1: Versicolor, 2: Virginica)]`  
- **Input Shape:** `(1, 4)`
- **Output Shape:** `(1,)`

```python
from sklearn.datasets import load_iris
data = load_iris()
X, y = data.data, data.target

# Model
model_3 = RandomForestClassifier(n_estimators=10)
model_3.fit(X, y)

# Save
joblib.dump(model_3, 'models/iris_classifier.pkl')

# Example usage
sample_input = [[5.1, 3.5, 1.4, 0.2]]
prediction = model_3.predict(sample_input)
print(f"Predicted Species ID: {prediction[0]} ({data.target_names[prediction[0]]})")
```

## 4. Sentiment Analysis Pipeline (NLP / Text)
**Domain:** Social Media / Marketing  
**Description:** Analyzes a string of text to determine if the sentiment is Positive or Negative.

### Schema
- **Input:** `string` (e.g., "I love this project!")
- **Output:** `[sentiment (0: Negative, 1: Positive)]`  
- **Note:** This model uses a Pipeline including a TF-IDF Vectorizer.
- **Input Shape:** `(1,)` list of strings
- **Output Shape:** `(1,)`

```python
texts = [
    "This is amazing work", "I hate this product", 
    "Excellent customer service", "Worst experience ever",
    "Highly recommended", "Total waste of money"
]
labels = [1, 0, 1, 0, 1, 0] # 1: Positive, 0: Negative

# Pipeline: Vectorizer + Model
model_4 = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('svc', SVC(probability=True))
])
model_4.fit(texts, labels)

# Save
joblib.dump(model_4, 'models/sentiment_analysis_model.pkl')

# Example usage
sample_input = ["This is the best tool I have ever used!"]
prediction = model_4.predict(sample_input)
print(f"Sentiment for '{sample_input[0]}': {'Positive' if prediction[0] == 1 else 'Negative'}")
```

## 5. Customer Segmentation (Clustering)
**Domain:** Retail / Marketing  
**Description:** Groups customers into clusters based on their spending habits and income.

### Schema
- **Input:** `[annual_income, spending_score]`
- **Output:** `[cluster_id (int)]`  
- **Input Shape:** `(1, 2)`
- **Output Shape:** `(1,)`

```python
X = np.array([[15, 39], [15, 81], [16, 6], [16, 77], [17, 40], [140, 97], [140, 10], [130, 80]])

# Model
model_5 = KMeans(n_clusters=3, n_init=10)
model_5.fit(X)

# Save
joblib.dump(model_5, 'models/customer_segmentation_model.pkl')

# Example usage
sample_input = [[20, 50]]
cluster = model_5.predict(sample_input)
print(f"Input {sample_input} assigned to Cluster: {cluster[0]}")
```

## 6. Daily Sales Forecaster (Regression)
**Domain:** Inventory / E-commerce  
**Description:** Predicts tomorrow's sales based on historical lags and seasonal features.

### Schema
- **Input:** `[sales_lag1, sales_lag2, sales_lag3, day_of_week, is_holiday]`
- **Output:** `[predicted_sales (float)]`  
- **Input Shape:** `(1, 5)`
- **Output Shape:** `(1,)`

```python
X = np.random.rand(100, 5) * 100
y = X[:, 0] * 0.5 + X[:, 1] * 0.3 + np.random.normal(0, 5, 100)

# Model
model_6 = GradientBoostingRegressor()
model_6.fit(X, y)

# Save
joblib.dump(model_6, 'models/sales_forecast_model.pkl')

# Example usage
sample_input = [[150.0, 145.0, 160.0, 3, 0]]
prediction = model_6.predict(sample_input)
print(f"Predicted Sales: {prediction[0]:.2f}")
```

## 7. Loan Approval Predictor (Deep Learning - MLP)
**Domain:** Banking / Fintech  
**Description:** Uses a neural network to determine if a loan application should be approved.

### Schema
- **Input:** `[credit_score, income, loan_amount, debt_to_income_ratio]`
- **Output:** `[approval_status (0: Reject, 1: Approve)]`  
- **Input Shape:** `(1, 4)`
- **Output Shape:** `(1,)`

```python
X = np.array([[700, 5000, 20000, 0.3], [550, 3000, 10000, 0.5], [800, 10000, 50000, 0.1], [600, 4000, 15000, 0.4]])
y = np.array([1, 0, 1, 0])

# Model
model_7 = MLPClassifier(hidden_layer_sizes=(10, 5), max_iter=1000)
model_7.fit(X, y)

# Save
joblib.dump(model_7, 'models/loan_approval_model.pkl')

# Example usage
sample_input = [[720, 6000, 25000, 0.25]]
prediction = model_7.predict(sample_input)
print(f"Loan Approval: {'Approved' if prediction[0] == 1 else 'Rejected'}")
```

## 10. Hand-written Digit Recognizer (KNN)
**Domain:** Computer Vision / OCR  
**Description:** Recognizes digits (0-9) from flattened 8x8 image pixels.

### Schema
- **Input:** `[pixel_values (64 values)]` (flattened 8x8 grayscale image)
- **Output:** `[digit (int: 0-9)]`  
- **Input Shape:** `(1, 64)`
- **Output Shape:** `(1,)`

```python
from sklearn.datasets import load_digits
digits = load_digits()
X, y = digits.data, digits.target

# Model
model_10 = KNeighborsClassifier(n_neighbors=3)
model_10.fit(X, y)

# Save
joblib.dump(model_10, 'models/digit_recognition_model.pkl')

# Example usage
sample_input = [X[0]]
prediction = model_10.predict(sample_input)
print(f"Recognized Digit: {prediction[0]}")
```
